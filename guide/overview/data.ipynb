## Understand the codes

A **study position** is a presentation's place in a trial; an **item code**
identifies what was presented there. A positive recall code `p` names study
position `p`; use `p - 1` to index its item code. Zero pads unused study and
response slots. These positions and within-list codes are distinct from the
global vocabulary IDs in `pres_itemids`.


````python
#| code-summary: "5 — Translate recall positions into within-list and global item IDs"

positions = data["recalls"][trial_idx]
positions = positions[positions > 0]
print("Recall positions:", positions)
print("Within-list item codes:", data["pres_itemnos"][trial_idx, positions - 1])
print("Global item IDs:", data["pres_itemids"][trial_idx, positions - 1])
````

For study codes `[7, 2, 7, 0]` and `listLength=3`, positions 1 and 3 contain
the same item (code 7); position 4 is unused. A recall code of 3 refers to
that item through its third-position presentation. Dataset simulation reports
the item's first-occurrence position, 1. A recall code alone therefore does
not tell us which presentation caused remembering.

Analyses can classify the named position or all presentations of its item.
See [Measurements and recall status](analyses.ipynb#measurements-and-recall-status)
for the distinction and the existing defaults.

## Additional fields

| Alignment | Examples | Keep aligned with |
|---|---|---|
| Trial | `session`, `list_type` | Trial rows. |
| Study event | `valence`, `pres_itemids`, measurements such as `LateLPP`, quality flags | Trial rows and study slots. |
| Recall event | `irt`, `rec_itemids` | Trial rows and response slots. |

A measurement and its quality flags occupy the same trial/position cells as
their presentations. A missing measurement or excluded observation does not
change the recorded recall codes. Vocabulary-wide features and distance
matrices remain separate inputs.

## Construct, save and reload

`make_dataset()` constructs the four core fields. The example below uses the selected trial's recorded participant and list length.

A single trial, participant ID or list length broadcasts to match the other
inputs. Participant IDs must be integers, supplied as integer scalars, sequences
or arrays; floating-point IDs such as `3.0` and Boolean IDs are outside the
interface. Study codes and lengths must be integer-valued. All core values must
be representable as `int32`. Construction converts supported inputs to arrays
without checking these preconditions. Supply compatible trial counts,
nonnegative lengths and zero padding beyond each declared length.

Supply `listLength` explicitly for padded study lists. Without it, the default
is the presentation array width. When presentations are omitted, lengths such
as `[2, 3]` generate sequential item codes, zero-padding the shorter list. If
both presentations and lengths are omitted, the constructor uses the response
width and highest recall code; this cannot establish the experimental study
length.

With explicit presentations, `make_dataset` works inside `jit` and `vmap`;
keep `reps` static. A static integer `listLength` also supports generated
presentations under JIT. Inferring the presentation width from array values
requires concrete data and should happen before compilation.


````python
#| code-summary: "6 — Construct one trial and save and reload its arrays"

one_trial = make_dataset(
    recalls=data["recalls"][trial_idx],
    pres_itemnos=data["pres_itemnos"][trial_idx],
    listLength=int(data["listLength"][trial_idx, 0]),
    subject=int(data["subject"][trial_idx, 0]),
)
with TemporaryDirectory() as folder:
    path = str(Path(folder) / "example.h5")
    save_dict_to_hdf5(one_trial, path)
    with h5py.File(path, "r") as stored:
        print("Stored recalls:", stored["data/recalls"].shape)
    reloaded = load_data(path)
    print("Loaded recalls:", reloaded["recalls"].shape)
    print("Arrays preserved:", all(
        bool(jnp.array_equal(one_trial[name], reloaded[name])) for name in one_trial
    ))
````

`save_dict_to_hdf5()` writes transposed arrays under `/data`; `load_data()` transposes them back into trial rows and returns JAX arrays. This is the helpers' storage convention, not a requirement of HDF5.

## Project paths and condition selections

These operations can be used in an ordinary Python script. Resolve the project
directory once, then keep file loading and example preparation explicit:

```python
from jaxcmr.helpers.paths import resolve_project_root
from jaxcmr.helpers.io import load_data
from jaxcmr.examples import fetch_inputs
from jaxcmr.dataset import select_conditions

root = resolve_project_root("~/my-research")
fetch_inputs(["data/Dupertuys2026.h5"], base_dir=root)
data = load_data(str(root / "data/Dupertuys2026.h5"))
conditions = select_conditions(
    data,
    [data["list_type"].ravel() == 1, data["list_type"].ravel() == 2],
    complete_lists_only=True,
)
for condition in conditions:
    print(condition.trial_count, condition.subject_ids)
```

`resolve_project_root(project_root=None)` honors an explicit path first. Otherwise,
it uses the current directory in Colab, the enclosing Git project elsewhere, or
the current directory when no Git project is found. It normalizes paths and
applies the existing platform checks; it does not create directories, change the
working directory, print, or fetch inputs. An explicit directory is useful for
scripts launched outside their research folder.

`select_conditions` accepts an ordered sequence of boolean trial masks, existing
query strings such as `"data['list_type'] == 1"`, or `None` for all trials. Each
`TrialSelection` contains a one-dimensional `mask`, `trial_count`, sorted unique
`subject_ids`, and a `participant_count` property. Pass its mask to an analysis;
the dataset itself is unchanged. Complete-list filtering excludes any trial with
nonpositive padding in `pres_itemnos`. Supply Boolean masks with one entry per
trial. Empty selections raise by default; set `require_nonempty=False` when
empty conditions are expected in your procedure. Query strings execute Python,
so use expressions you trust.
