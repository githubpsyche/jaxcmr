Computational models of memory describe how people learn and retrieve information. Implementing these models and comparing their predictions with experimental data can require substantial programming.

jaxcmr is a Python package for computational memory research. It leverages a [compositional design](https://refactoring.guru/design-patterns/bridge) and [JAX](https://docs.jax.dev/en/latest/)'s support for just-in-time compilation, vectorization, and automatic differentiation to streamline data analysis and model development so researchers can explore more ideas in less time.

If you use jaxcmr, please cite the paper [citation to be added].

See the code repository for version [release notes](https://github.com/githubpsyche/jaxcmr/releases).
